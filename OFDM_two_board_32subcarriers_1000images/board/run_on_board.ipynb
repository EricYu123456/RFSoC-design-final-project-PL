# RFSoC 4×2 四天線批次收發

A 的 DAC_A 對 B 的 ADC_A 傳圖，B 的 DAC_A 對 A 的 ADC_A 回覆。兩端為 500 MHz 同頻分時。每張 300 包，每個視窗至多 8 包。

## 分別在兩板的 Jupyter terminal 執行

先在 B 執行 `sudo -E python3 run_on_board.py receive`。等到 RX_WAITING_FOR_PILOT，在 A 執行 `sudo -E python3 run_on_board.py send --input-dir data --count 1000`。首次驗證可將 count 改成 1。完成後從 B 執行下列儲存格顯示收到的圖。

```python
from pathlib import Path
import json
from IPython.display import display, Image
root = Path('rx_results')
result = json.loads((root / 'result.json').read_text())
frame = 0
folder = root / format(result['session'], '08x') / f'{frame:04d}'
display(Image(filename=str(folder / 'received.png')))
print(result['frames'][frame])
```

將兩板結果複製到同一台電腦，用 `python run_on_board.py verify --tx-dir tx_results --rx-dir rx_results --output two_board_result.json` 確認每張圖。
